### **A1.** Uma variável registra o **nível de satisfação do cliente**, em uma escala de 0 a 10. Em que nível de mensuração ela está? **O que isso permite** calcular e o que **não** permite?

A satisfação de 0 a 10 está no nível ordinal, pois os valores indicam ordem de satisfação, mas não asseguram distâncias iguais entre notas. Permite calcular frequências, percentuais, moda e mediana, e ordenar as respostas. Não permite, rigorosamente, calcular razões ou afirmar que uma nota 8 representa o dobro da satisfação de uma nota 4.


### **A2.** Aplique o **teste do zero** à variável **“número de funcionários de uma empresa”** e diga em que nível ela está.

No teste do zero, “zero funcionários” significa ausência de funcionários, logo, o zero é absoluto, e não uma referência arbitrária. Assim, a variável está no nível de razão, permitindo afirmar, por exemplo, que uma empresa com 20 funcionários tem o dobro de funcionários de outra com 10

---


### **A3.** Você tem a **receita anual** de 449 companhias abertas e precisa escolher **uma** medida para representar o conjunto. Qual você escolhe e por quê?

Eu escolheria a mediana, pois receitas muito elevadas de algumas companhias podem puxar a média para cima, enquanto a mediana é menos sensível a valores extremos e representa melhor uma empresa típica nesse cenário. Ordenando as 449 receitas, ela corresponde ao 225º valor, com 224 observações de cada lado.

---


### **A4.** Por que **somar os desvios** de cada valor em relação à média **não** serve como medida de dispersão? E por que, então, se eleva ao quadrado?

A soma dos desvios em relação à média é sempre zero, porque os desvios positivos e negativos se cancelam, independentemente da dispersão dos dados. Ao elevá-los ao quadrado, os termos ficam não negativos e deixam de se cancelar, permitindo medir a dispersão e dando origem à variância
---


### **A5.** Você precisa comparar a **dispersão do preço da gasolina**, medido em R\$ por litro, com a **dispersão do IPCA**, medido em pontos percentuais. Que medida você usa e por quê?

Usaria o coeficiente de variação, calculado como (desvio-padrão ÷ média) × 100%, pois permite comparar a dispersão relativa de variáveis com unidades diferentes. No caso do IPCA, é preciso cautela: se a média estiver próxima de zero ou houver valores negativos, essa medida pode ser inadequada ou enganosa
---


### **A6.** A média de uma variável é R\$ 6,13 e a mediana é R\$ 6,08. **O que essa diferença sugere** sobre a forma da distribuição — e o que ela **não** permite concluir?

A média maior que a mediana sugere assimetria à direita, com valores mais altos puxando a média para cima. Porém, essa diferença de R$ 0,05, isoladamente, não comprova a forma da distribuição, não permite avaliar a intensidade da assimetria nem afirmar que existem valores extremos.

### **A7.** Uma empresa diz: *“a inadimplência média do período foi de 3,54%, e o valor de hoje é 3,54%”*. O que está errado nessa afirmação?

Não há necessariamente erro: a média do período e o valor de hoje podem coincidir em 3,54%, embora representem informações diferentes. O erro seria concluir que a inadimplência de hoje é 3,54% apenas porque essa foi a média do período, sem verificar o valor atual.

### B2. A medida de posição e a justificativa

A variável `receita` está em milhares de reais. Calcule **as três medidas de posição** e
decida **qual delas** descreve melhor o conjunto. **Justifique com os números.**

```python
# CÉLULA B2: as três medidas de posição da receita.
receita = base["receita"].dropna()
print(f"n = {len(receita)}")
print(f"média   = {receita.mean():,.0f} milhares de reais".replace(",", "."))
print(f"mediana = {receita.median():,.0f} milhares de reais".replace(",", "."))
print(f"razão média/mediana = {receita.mean() / receita.median():.1f} vezes")
```

### B3. A dispersão

Calcule o **desvio padrão**, o **coeficiente de variação** e o **IQR** da receita.
Depois, escreva **uma frase** que diga o que esses números mostram.

```python
# CÉLULA B3: a dispersão da receita.
desvio = receita.std()
cv = desvio / receita.mean() * 100
q1, q3 = receita.quantile(0.25), receita.quantile(0.75)
print(f"desvio padrão = {desvio:,.0f}".replace(",", "."))
print(f"coeficiente de variação = {cv:.2f}%")
print(f"IQR = {q3 - q1:,.0f}   (Q1 {q1:,.0f} | Q3 {q3:,.0f})".replace(",", "."))
print()
print("O CV acima de 100% é o sinal numérico da assimetria:")
print("o desvio é maior que a própria média.")
print()
print("Conferência (o valor esperado está no gabarito do professor):")
```

A receita apresenta alta dispersão, com desvio padrão de 40.915.645 e coeficiente de variação de 331,33%, indicando grande diferença entre os valores. Além disso, o intervalo interquartil de 7.168.750 mostra que há uma distância considerável entre os 50% centrais das receitas, reforçando a desigualdade entre as empresas.

### B4. A forma da distribuição

Faça o **histograma** da receita e descreva a forma em **uma frase, podendo usar
números**.

```python
# CÉLULA B4: o histograma da receita.
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.hist(receita, bins=30, color="#0AB0AB", edgecolor="white")
ax.axvline(receita.mean(), color="#8D0333", linewidth=2, label="média")
ax.axvline(receita.median(), color="#D4B277", linewidth=2, linestyle="--", label="mediana")
ax.set_xlabel("Receita (milhares de reais)")
ax.set_ylabel("Número de companhias")
ax.set_title("Receita das companhias abertas, 2024")
ax.legend()
plt.tight_layout()
plt.show()

print(f"assimetria calculada: {receita.skew():.2f}")
```

O histograma apresenta uma distribuição assimétrica à direita, com assimetria de 7,56, indicando que a maioria das empresas possui receitas menores e poucas apresentam receitas extremamente elevadas, puxando a média para a direita.

**5. A base escolhida e a célula que a carrega**

_(cole abaixo o código que carrega a sua base e execute-o. A fonte pode ser um arquivo
publicado ou uma **API** — o que importa é que a célula rode sem erro)_


```python
import io
import zipfile
import requests
import pandas as pd

url = "https://cetic.br/media/microdados/907/tic_empresas_2024_p%C3%B3s_cec_base_de_microdados.zip"

# Baixar o arquivo
resposta = requests.get(url, timeout=180)
resposta.raise_for_status()

# Abrir o ZIP e selecionar o arquivo de dados
with zipfile.ZipFile(io.BytesIO(resposta.content)) as pacote:
    arquivos = [
        item for item in pacote.infolist()
        if not item.is_dir()
        and "__MACOSX/" not in item.filename
        and not item.filename.split("/")[-1].startswith(".")
    ]

    csvs = [
        item for item in arquivos
        if item.filename.lower().endswith(".csv")
    ]

    planilhas = [
        item for item in arquivos
        if item.filename.lower().endswith((".xlsx", ".xls"))
    ]

    candidatos = csvs or planilhas

    if not candidatos:
        raise ValueError(
            "O ZIP não contém CSV ou Excel. Arquivos encontrados:\n"
            + "\n".join(item.filename for item in arquivos)
        )

    escolhido = max(candidatos, key=lambda item: item.file_size)
    conteudo = pacote.read(escolhido)

# Ler a base
if escolhido.filename.lower().endswith(".csv"):
    for codificacao in ("utf-8-sig", "cp1252", "latin-1"):
        try:
            texto = conteudo.decode(codificacao)
            break
        except UnicodeDecodeError:
            continue

    minha_base = pd.read_csv(
        io.StringIO(texto),
        sep=None,
        engine="python",
        decimal=","
    )
else:
    minha_base = pd.read_excel(io.BytesIO(conteudo))

print("Arquivo selecionado:", escolhido.filename)
print("Quantidade de linhas:", minha_base.shape[0])
print("Quantidade de colunas:", minha_base.shape[1])

minha_base.head()
```
